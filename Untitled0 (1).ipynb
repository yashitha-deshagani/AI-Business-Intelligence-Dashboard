```python
import pandas as pd
df = pd.read_excel("sample_-_superstore.csv")
df.info()
```

```python
df.duplicated().sum()
```

```python
df[['Sales', 'Quantity', 'Discount', 'Profit']].describe()
```

```python
df.groupby('Category')[['Sales', 'Profit']].sum().sort_values('Sales', ascending=False)
```

```python
df.groupby('Category')['Discount'].mean().sort_values(ascending=False)
```

**Insight 1:** Furniture generates nearly as much revenue as Office Supplies (\$755K vs \$732K) but only ~3% profit margin compared to Office Supplies' ~17%. Furniture does have the highest average discount rate (17.3% vs 13.1% for Technology), but the discount gap alone doesn't fully explain the margin gap, suggesting Furniture also carries higher costs relative to price.

```python
df.groupby('Sub-Category')['Profit'].sum().sort_values()
```

**Insight 2:** Tables are the single biggest profit problem in the business, losing \$17,753 overall - more than 4x the losses of the next-worst sub-category (Bookcases, -\$3,632). This explains why Furniture's overall margin is so thin (Insight 1): its profitable sub-categories (Chairs, Furnishings) are being dragged down by Tables and Bookcases. Copiers (\$56K) and Phones (\$45K) are the most profitable sub-categories overall.

```python
df[df['Sub-Category'] == 'Tables']['Discount'].mean()
```

**Insight 3:** Tables are discounted at 25.8% on average - over 10 points higher than the dataset average (~15.5%) and the highest of any sub-category checked so far. This heavy discounting is the most likely direct cause of Tables' -\$17,753 loss, suggesting the business should reduce discount depth on Tables or renegotiate their cost basis.

```python
df.set_index('Order Date').resample('ME')['Sales'].sum()
```

**Insight 4:** Sales show strong, repeating seasonality - January and February are consistently the weakest months every year, while September, November, and December are consistently the strongest, pointing to a clear holiday/back-to-school buying pattern. Overall sales also show a year-over-year upward trend, with November 2026 (\$118.5K) far outpacing November in prior years (~\$76-79K).

```python
df.groupby('Customer Name')['Profit'].sum().sort_values(ascending=False).head(10)
```

**Insight 5:** Tamara Chand is the most valuable customer by far, generating \$8,981 in profit - 29% more than the next-highest customer (Raymond Buch, \$6,976). The top 10 customers alone account for a significant share of total profit, suggesting the business would benefit from a customer retention/loyalty program focused on this group.

```python
import sqlite3
conn = sqlite3.connect(':memory:')
df.to_sql('superstore', conn, index=False, if_exists='replace')
```

```python
query = """
SELECT "Product Name", SUM(Sales) AS total_revenue
FROM superstore
GROUP BY "Product Name"
ORDER BY total_revenue DESC
LIMIT 10
"""
pd.read_sql_query(query, conn)
```

First query answers the question of which products generate the most revenue.

```python
query = """
SELECT Region, SUM(Sales) AS total_revenue, SUM(Profit) AS total_profit
FROM superstore
GROUP BY Region
ORDER BY total_profit DESC
"""
pd.read_sql_query(query, conn)
```

Second query answers the question of which region performs the best.

**SQL Insight 1:** West is the top-performing region by both revenue (\$739.8K) and profit (\$110.8K). Central generates the 2nd-highest revenue (\$503.2K) but converts it into the least profit of any region (\$39.9K) - a pattern similar to Furniture's weak margin, worth investigating further (e.g. discount rates by region).

```python
query = """
SELECT "Customer Name", SUM(Profit) AS total_profit
FROM superstore
GROUP BY "Customer Name"
ORDER BY total_profit DESC
LIMIT 10
"""
pd.read_sql_query(query, conn)
```

Third query answers the question of who the highest value customers are.

```python
query = """
SELECT strftime('%Y-%m', "Order Date") AS month, SUM(Sales) AS total_revenue
FROM superstore
GROUP BY month
ORDER BY month
"""
pd.read_sql_query(query, conn)
```

Fourth query answers the question of what the monthly sales trends are.

```python
query = """
SELECT "Sub-Category", SUM(Profit) AS total_profit
FROM superstore
GROUP BY "Sub-Category"
HAVING total_profit < 0
ORDER BY total_profit
"""
pd.read_sql_query(query, conn)
```

Fifth query answers the question of which products are underperforming.

**SQL Insight 2:** Using SQL's HAVING clause to filter aggregated results, the only sub-categories with negative total profit are Tables (-\$17,753), Bookcases (-\$3,632), and Supplies (-\$1,171) - confirming the same underperformers found in the Pandas analysis.
